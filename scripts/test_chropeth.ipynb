```python
import json
from urllib.request import urlopen

import plotly.graph_objects as go

with urlopen(
    "https://raw.githubusercontent.com/plotly/datasets/master/geojson-counties-fips.json"
) as response:
    counties = json.load(response)

import pandas as pd

df = pd.read_csv(
    "https://raw.githubusercontent.com/plotly/datasets/master/fips-unemp-16.csv",
    dtype={"fips": str},
)


fig = go.Figure(
    go.Choroplethmap(
        geojson=counties,
        locations=df.fips,
        z=df.unemp,
        colorscale="Viridis",
        zmin=0,
        zmax=24,
        marker_opacity=0.5,
        marker_line_width=0,
    )
)
fig.update_layout(
    map_style="carto-positron", map_zoom=3, map_center={"lat": 37.0902, "lon": -95.7129}
)
fig.update_layout(margin={"r": 0, "t": 0, "l": 0, "b": 0})
fig.show()
```

```python
import geopandas as gpd
import plotly.express as px

df = px.data.election()
geo_df = (
    gpd.GeoDataFrame.from_features(px.data.election_geojson()["features"])
    .merge(df, on="district")
    .set_index("district")
)

fig = px.choropleth_map(
    geo_df,
    geojson=geo_df.geometry,
    locations=geo_df.index,
    color="Joly",
    center={"lat": 45.5517, "lon": -73.7073},
    map_style=None,
    zoom=8.5,
)
fig.show()
```

```python
df_car_share = px.data.carshare()
df_car_share.head()
```

```python
import numpy as np
import plotly.express as px
import plotly.figure_factory as ff

fig = ff.create_hexbin_map(
    data_frame=df_car_share,
    lat="centroid_lat",
    lon="centroid_lon",
    nx_hexagon=4,
    opacity=0.2,
    color="peak_hour",
    agg_func=np.sum,
)
fig.update_layout(margin=dict(b=0, t=0, l=0, r=0))
fig.show()
```
